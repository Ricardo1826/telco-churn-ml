# Nettoyage et prétraitement : Telco Customer Churn

## Objectif
Charger le dataset brut, explorer sa structure, traiter les valeurs manquantes et incohérences, préparer les données pour l'analyse exploratoire et la modélisation.

## Étapes
1. Chargement du dataset
2. Aperçu général (dimensions, types, valeurs manquantes)
3. Vérification des doublons
4. Traitement des incohérences de type
5. Export du dataset nettoyé

```python
import pandas as pd
import numpy as np

df = pd.read_csv('../data/raw/WA_Fn-UseC_-Telco-Customer-Churn.csv')
print(df.shape)
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df[pd.to_numeric(df['TotalCharges'], errors='coerce').isnull()]
```

```python
df[pd.to_numeric(df['TotalCharges'], errors='coerce').isnull()][['tenure', 'MonthlyCharges', 'TotalCharges']]
```

```python
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')
df['TotalCharges'] = df['TotalCharges'].fillna(0)

print(df['TotalCharges'].isnull().sum(), 'valeurs manquantes restantes')
print(df['TotalCharges'].dtype)
```

```python
df['TotalCharges_estime'] = df['tenure'] * df['MonthlyCharges']
df[['TotalCharges', 'TotalCharges_estime']].corr()
```

```python
df = df.drop(columns=['TotalCharges_estime'])
```

```python
print(df.duplicated().sum(), 'lignes dupliquées')
print(df['customerID'].duplicated().sum(), 'customerID dupliqués')
```

```python
df['SeniorCitizen'].unique()
```

```python
df['SeniorCitizen'] = df['SeniorCitizen'].map({0: 'No', 1: 'Yes'})
df['SeniorCitizen'].unique()
```

```python
df.to_csv('../data/processed/telco_cleaned.csv', index=False)
print('Fichier exporté avec succès :', df.shape)
```
